```python
categorical_columns = cleaned_df.select_dtypes(include=['object']).columns
categorical_columns
```

```python
for cols in categorical_columns :
  display(cleaned_df[cols].value_counts())
  print('==================================')
```

```python
numerical_columns = cleaned_df.select_dtypes(include=['int64', 'float64']).columns
numerical_columns
```

```python
for ncol in numerical_columns:
    fig = px.histogram(cleaned_df, x=ncol, nbins=30, title=f"Distribution of {ncol}")
    fig.show()
```

```python
for ncol in numerical_columns:
    plt.figure()
    sns.histplot(data=cleaned_df, x=ncol, kde=True)
    plt.show()

```

```python
for ncol in numerical_columns:
    plt.figure()
    sns.kdeplot(data=cleaned_df, x=ncol)
    plt.show()
```

```python
#box plot
for ncol in numerical_columns:
    plt.figure()
    sns.boxplot(data=cleaned_df, x=ncol)
    plt.show()
```

```python
#vilion plot
for ncol in numerical_columns:
    plt.figure()
    sns.violinplot(data=cleaned_df, x=ncol)
    plt.show()
```

```python
for ncol in numerical_columns:
    plt.figure()
    sns.violinplot(data=cleaned_df, x=ncol, y='gender', hue='gender', palette={
        'Male':'blue',
        'Female':'pink'
    })
    sns.pointplot(
        data=cleaned_df,
        x=ncol,
        y='gender',
        estimator=np.mean,
        markers='D',
        color='red'
    )
    plt.show()
```

```python



for ccol in categorical_columns:
    counts = cleaned_df[ccol].value_counts().reset_index()
    counts.columns = [ccol, "count"]   # rename columns properly

    fig = px.bar(
        counts,
        x=ccol,
        y="count",
        text="count",
        title=f"Count of categories in {ccol}",
        labels={ccol: ccol, "count": "Count"}
    )
    fig.update_traces(textposition="outside", marker_line_width=1, marker_line_color="black")
    fig.update_layout(bargap=0.2)
    fig.show()
```

```python
#pie chart
for ccol in categorical_columns:
    counts = cleaned_df[ccol].value_counts().reset_index()
    counts.columns = [ccol, "count"]

    fig = px.pie(
        counts,
        names=ccol,
        values="count",
        title=f"Distribution of {ccol}",
        hole=0.3,
        labels={ccol: ccol, "count": "Count"}
    )
    fig.show()
```

```python
#heatmap
corr_matrix = cleaned_df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
```

```python
#scatterplot monthly charges vs total charges
sns.scatterplot(data=cleaned_df, x='MonthlyCharges', y='TotalCharges')
plt.title('Monthly Charges vs Total Charges')
plt.xlabel('Monthly Charges')
plt.ylabel('Total Charges')
```
